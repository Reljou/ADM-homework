```python
# sWAP cASE

def swap_case(s):
    new_string=''
    for i in range(len(s)):
        if s[i].isalpha():
            if s[i]==s[i].upper():
                new_string+=s[i].lower()
                continue
            if s[i]==s[i].lower():
                new_string+=s[i].upper()
                continue
        new_string+=s[i].lower()
    return new_string

if __name__ == '__main__':
    s = input()
    result = swap_case(s)
    print(result)
```

```python
# Find a string

def count_substring(string, sub_string):
    p=len(sub_string)
    x=len(string)-len(sub_string)
    counter=0
    for i in range(x+1):
        if string[i:i+p]==sub_string:
            counter+=1
    return counter

if __name__ == '__main__':
    string = input().strip()
    sub_string = input().strip()
    
    count = count_substring(string, sub_string)
    print(count)
```

```python
# String Validators 
def has_alnum(s):
    for i in range(len(s)):
        if s[i].isalnum():
            return True
        if i==(len(s)-1):
            return False
def has_alpha(s):
    for i in range(len(s)):
        if s[i].isalpha():
            return True
        if i==(len(s)-1):
            return False
def has_digit(s):
    for i in range(len(s)):
        if s[i].isdigit():
            return True
        if i==(len(s)-1):
            return False
def has_lower(s):
    for i in range(len(s)):
        if s[i].islower():
            return True
        if i==(len(s)-1):
            return False
def has_upper(s):
    for i in range(len(s)):

        if s[i].isupper():
            return True
        if i==(len(s)-1):
            return False
if __name__ == '__main__':
    s = input()
    print(has_digit(s))
    print(has_lower(s))
    print(has_upper(s))
```

```python
# Text Wrap

def wrap(string, max_width):
    count =0 
    paragraph =""
    for i in string:
        count =count+ 1
        if count == max_width:
            paragraph+=i
            paragraph+="\n"
            count =0
        else:
            paragraph+=i
    return paragraph
if __name__ == '__main__':
    string, max_width = input(), int(input())
    result = wrap(string, max_width)
    print(result)
```

```python
# Find The Maximum Depth

import xml.etree.ElementTree as etree

maxdepth = 0
def depth(node, level):
    global maxdepth
    maxdepth = max(maxdepth, level + 1)
    for child in node:
        depth(child, level + 1)  

if __name__ == '__main__':
    n = int(input())
    xml = ""
    for i in range(n):
        xml =  xml + input() + "\n"
    tree = etree.ElementTree(etree.fromstring(xml))
    depth(tree.getroot(), -1)
    print(maxdepth)
```

```python
# Decorators 2

import operator
def person_lister(f):
    def inner(people):
        for x in people:
            x[2] = int(x[2])
        people.sort(key=operator.itemgetter(2))
        return [f(x) for x in people]
    return inner

@person_lister
def name_format(person):
    return ("Mr. " if person[3] == "M" else "Ms. ") + person[0] + " " + person[1]

if __name__ == '__main__':
    people = [input().split() for i in range(int(input()))]
    print(*name_format(people), sep='\n')
```
